## Conditionals, loops and functions

```python
mark = 45
grade = None
message = None

if mark < 40:
    grade = 'Fail'
    message = 'Please try again.'
elif mark < 50:
    grade = 'Pass'
    message = 'Well done for passing.'
elif mark < 60:
    grade = '2:2'
    message = 'Good.'
elif mark < 70:
    grade = '2:1'
    message = 'Very good.'
else:
    grade = '1st'
    message = 'Excellent!'

output=f'Your mark was {mark}, your grade was {grade}. {message}'

print(output)
```

```python
def eligible_for_snake_wrestling(snake,age,venomous,length,number_wrestles,number_wins):

    # Set the initial value of eligible - could include more conditions here
    eligible = (5 <= age < 20) and (not venomous)
    
    # Must be a snake
    eligible = eligible and snake
    
    # Must be more than 1 and no more than 2 metres long
    eligible = eligible and (1 < length <= 2)
    
    # Entrants that have wrestled before should only have won less than half their matches
    if number_wrestles > 0:
        eligible = eligible and (number_wins < number_wrestles / 2)
    
    # Entrants that have wrestled more than five times should be no more than 12 years old
    if number_wrestles > 5:
        eligible = eligible and (age <= 12)
    
    return eligible
```

```python
# Variables to test the function
snake = True
name = 'Boomslang'
age = 13
venomous = True
length = 1.5
number_wrestles = 0
number_wins = 0

eligible=eligible_for_snake_wrestling(snake,age,venomous,length,number_wrestles,number_wins)

if eligible:
    print( name, "is eligible :)")
else:
    print( name, "is not eligible :(")
```

```python
# Your program will be tested on the following data:
mailingList = [
    ['title', 'fname', 'sname', 'addr1', 'addr2', 'postcode'],
    ['Mrs', 'Joan', 'Smith', '3 Huntsmans Avenue', 'Batley', 'WF17 3RW'],
    ['Mr', 'Fred', 'Jones', '15a Brighton Road', 'Wyke', 'BD6 4NN'],
    ['Dr', 'Maria', 'Tan', '122 High Street', 'Ilkley', 'LS29 2AD']
]
```

```python
def formatName(title, fname, sname):
    # Format as Title F. Surname
    return f'{title} {fname[0]}.{sname}'

def formatEntry(entry):
    title, fname, sname, addr1, addr2, postcode = entry
    name_line = formatName(title, fname, sname)
    # Return a list of three strings: name, address line, town, postcode
    return [name_line, addr1, addr2, postcode]


def prettyPrintAddressLabels(recipients):
    # Loop through the records in the mailing list and format the addresses
    formatted = []
    for entry in recipients[1:]:  # Skip the header row
        formatted.append(formatEntry(entry))
    # return a list of formatted entries
    return formatted

# Test the function
results = prettyPrintAddressLabels(mailingList)
for label in results:
    for line in label:
        print(line)
    print()  # Blank line between labels
```
